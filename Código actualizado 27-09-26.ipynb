```python
# Se importan las librerias necesarias
import matplotlib.pyplot as plt  # Matplotlib permite graficar
import numpy as np  # Numpy permite trabajar con arreglos numericos
# Se define la funcioncuadratica
def f(x):
  return -x**2 + 6 * x + 7  # x**2 significa x elevado al cuadrado
# Se crea 300 Valores de x entre -1 y 10
x = np.linspace(-1, 10, 300)
# Se calcula los valores de la funcion para cadad valor de x
y = f(x)
# graficamos la funcion
# label para poner el nombre de la curva en la leyenda. color='red'la curva se dibuja de color rojo
plt.plot(x, y, label="W(t)=-t^2+6t+7", color="red")
# Se dibuja el eje horizontal y el eje vertical en x=0 y en y=0 con color gris y lineas punteadas
plt.axhline(0, color="gray", linestyle="--")
plt.axvline(0, color="gray", linestyle="--")
# se maraca las raices de la funcion
plt.scatter(-1, 0, label="Raíz(-1,0)")
plt.scatter(7, 0, label="Raíz(7,0)")
# se marca el corte con el eje y
plt.scatter(0, 7, label="Corte eje (0,7)")
# se marca el vertice de laparabola
plt.scatter(3, 16, label="Vértice (3,16)")
# se dibuja el eje de simeria de la parabola en x=3 con color verde y dashdot indica con lineas y puntos
plt.axvline(3, linestyle="dashdot", label="eje de simetria t=3", color="green",)
# agregamos titulo y nombres de los ejes
plt.title("Funcion Cuadratica")
plt.xlabel("t")
plt.ylabel("W(t)")
# se activa la cuadricula y la leyenda
plt.grid()
plt.legend()
# se muestra la grafica
plt.show()
```
